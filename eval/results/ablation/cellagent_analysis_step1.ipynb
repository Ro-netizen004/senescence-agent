```python
import scanpy as sc
import anndata as ad

adata = sc.read_h5ad("cellagent_null.h5ad")

if 'group' in adata.obs.columns:
    print("Column 'group' exists in adata.obs.")
    group_categories = adata.obs['group'].unique()
    if 'groupA' in group_categories and 'groupB' in group_categories:
        print("Categories 'groupA' and 'groupB' are present in 'group' column.")
        print("Cell counts per group:")
        print(adata.obs['group'].value_counts())
    else:
        print("Categories 'groupA' or 'groupB' are missing in 'group' column.")
        print("Available categories:", group_categories)
else:
    print("Column 'group' does not exist in adata.obs.")
```
