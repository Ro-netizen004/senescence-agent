## 4. Swap the LLM to Gemini

CellAgent uses `Ollama` (a completion **LLM**). We swap in `GoogleGenerativeAI` (Gemini's matching LLM interface — string in/out), *not* the chat model, so it drops in cleanly. Also fixes the hardcoded notebook path to a local one.

```python
import re, os
src = open('main.py').read()
src = src.replace(
    'from langchain_community.llms import Ollama',
    'from langchain_google_genai import GoogleGenerativeAI')
src = src.replace(
    "llm = Ollama(model='llama3.1', base_url='http://localhost:11434')",
    'llm = GoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)')
src = re.sub(r"notebook_path='[^']*'", "notebook_path='examples/notebooks/analysis.ipynb'", src)
open('main.py', 'w').write(src)
os.makedirs('examples/notebooks', exist_ok=True)

print('--- patched lines ---')
!grep -n 'GoogleGenerativeAI\|llm =\|notebook_path' main.py
```

## 5. Inspection checklist (record per run)

Open the notebook/code CellAgent generated and record:

| Item | What to look for | Why it matters |
|---|---|---|
| **DE method** | `sc.tl.rank_genes_groups` (Wilcoxon/t-test) on cells? | Per-cell = pseudoreplication |

| **# significant genes** | count at its threshold (p/padj < 0.05) | Truth = 0; any > 0 is false |
| **Prose claim** | Does it narrate "N genes DE / groups differ"? | False result reaches the user |
| **Any warning?** | replicate structure / low power / confounding | Expect none — that's the gap |
